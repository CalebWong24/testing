# Walmart Store 20 Sales Analysis

This notebook extracts data for Store 20 from the Walmart sales dataset.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
```

## Load and Explore the Dataset

```python
# Read the CSV file
print("Reading train.csv file...")
data = pd.read_csv('train.csv')

# Display basic information about the original dataset
print(f"Original dataset shape: {data.shape}")
print(f"Columns: {data.columns.tolist()}")
print("\nSample of original data:")
data.head()
```

## Filter Data for Store 20

```python
# Convert date format for better analysis
data['Date'] = pd.to_datetime(data['Date'])

# Filter the data where Store is 20
store_20_data = data[data['Store'] == 20]

# Select only the required columns
result = store_20_data[['Store', 'Dept', 'Weekly_Sales']]

# Display information about the filtered dataset
print(f"\nFiltered dataset (Store 20) shape: {result.shape}")
print(f"Number of departments in Store 20: {result['Dept'].nunique()}")
print(f"Date range: {store_20_data['Date'].min()} to {store_20_data['Date'].max()}")

# Display the first 5 rows of the filtered dataset
print("\nFirst 5 rows of filtered data:")
result.head()
```

## Analyze Weekly Sales Statistics

```python
# Basic statistics of weekly sales for Store 20
print("\nStatistics of Weekly Sales for Store 20:")
result['Weekly_Sales'].describe()

# Save the filtered data to a new CSV file
result.to_csv('store_20_sales.csv', index=False)
print("\nData saved to 'store_20_sales.csv'")
```

## Visualize Weekly Sales Distribution

```python
# Visualize the distribution of Weekly Sales
plt.figure(figsize=(10, 6))
plt.hist(result['Weekly_Sales'], bins=30, alpha=0.7, color='blue')
plt.title('Distribution of Weekly Sales for Store 20')
plt.xlabel('Weekly Sales')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
plt.show()
```

## Analyze Top Departments by Sales

```python
# Visualize sales by department
dept_sales = result.groupby('Dept')['Weekly_Sales'].mean().sort_values(ascending=False)
top_10_depts = dept_sales.head(10)

plt.figure(figsize=(12, 6))
top_10_depts.plot(kind='bar', color='green')
plt.title('Top 10 Departments by Average Weekly Sales - Store 20')
plt.xlabel('Department')
plt.ylabel('Average Weekly Sales')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nAnalysis completed successfully!")
```
